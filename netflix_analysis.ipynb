## Bước 1: Tổng quan dữ liệu
Yêu cầu: Đọc file `netflix_titles.csv` vào DataFrame `df`. Cho biết số dòng, số cột, và kiểu dữ liệu của từng cột.

```python
import pandas as pd

# Đọc file dữ liệu bằng đường dẫn tuyệt đối để tránh lỗi sai thư mục (FileNotFoundError)
df = pd.read_csv(r'D:\kada\backend\netflix_titles.csv')

# Lấy số dòng, số cột
so_dong, so_cot = df.shape
print(f"Số dòng: {so_dong} | Số cột: {so_cot}\n")

# Xem kiểu dữ liệu của từng cột
print("Kiểu dữ liệu của từng cột:")
print(df.dtypes)
```

## Bước 2: Tìm kiếm các lỗi dữ liệu cụ thể
Yêu cầu: Kiểm tra xem có dòng nào bị trùng lặp hoàn toàn, có `show_id` nào trùng hay không, và tìm hiểu xem có bao nhiêu dữ liệu bị thiếu (missing) cũng như kiểu dữ liệu thực tế của cột ngày tháng `date_added`.

```python
# Kiểm tra dòng trùng lặp hoàn toàn
so_dong_trung = df.duplicated().sum()
print(f"Số dòng trùng lặp hoàn toàn: {so_dong_trung}")

# Kiểm tra show_id trùng lặp
so_id_trung = df['show_id'].duplicated().sum()
print(f"Số show_id bị trùng lặp: {so_id_trung}\n")

# Kiểm tra dữ liệu bị thiếu ở từng cột
print("Số lượng giá trị bị thiếu ở mỗi cột:")
print(df.isnull().sum())

# Xem xét kiểu dữ liệu của date_added
print(f"\nKiểu dữ liệu hiện tại của 'date_added': {df['date_added'].dtype} (kiểu chuỗi/object cần được đổi thành datetime)")
```

## Bước 3: Đề xuất, áp dụng cách sửa lỗi và kiểm tra lại
Yêu cầu:
1. Chuyển đổi cột `date_added` sang định dạng thời gian đúng (datetime).
2. Xử lý dữ liệu thiếu: Điền `Unknown` cho các cột `director`, `cast`, `country` vì không có thông tin.
3. Sau khi sửa, kiểm tra lại kiểu dữ liệu và đếm số lượng lỗi còn lại.

```python
# 1. Chuyển đổi cột 'date_added' sang kiểu datetime
# Tham số errors='coerce' giúp chuyển các giá trị không hợp lệ thành NaT
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')

# 2. Điền 'Unknown' cho các cột có nhiều giá trị thiếu
cot_can_dien = ['director', 'cast', 'country']
df[cot_can_dien] = df[cot_can_dien].fillna('Unknown')

# 3. Kiểm tra lại dữ liệu sau khi sửa
print("Kiểu dữ liệu sau khi chuyển đổi:")
print(df.dtypes)

print("\nSố lượng giá trị bị thiếu hiện tại (những cột trên đã hết lỗi missing):")
print(df.isnull().sum())
```

## Bước 4: Lưu dữ liệu đã làm sạch sang file mới
Yêu cầu: Cho đến nay, `df` mới chỉ được làm sạch trên bộ nhớ của kernel. Lưu `df` đã làm sạch sang một file mới tên là `netflix_titles_clean.csv` trong cùng thư mục với file ban đầu. Không ghi đè lên file gốc `netflix_titles.csv`.

```python
# Lưu dataframe df vào file mới. 
# Tham số index=False giúp không sinh thêm một cột stt (index) thừa vào file CSV.
df.to_csv(r'D:\kada\backend\netflix_titles_clean.csv', index=False)

print("Đã lưu thành công dữ liệu sạch ra file 'netflix_titles_clean.csv'!")
```
